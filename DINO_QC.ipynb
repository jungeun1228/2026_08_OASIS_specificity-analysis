# Qualitative QC

# PCA - (1) raw concatenate (2) channel-wise L2 normalized

```python
feature_cols = [c for c in df_dino.columns if not c.startswith("Metadata")]
channels = ['AGP', 'Brightfield', 'DNA', 'ER', 'Mito', 'RNA']

X_raw = df_dino[feature_cols].to_numpy()
```

```python
import numpy as np

X_l2_parts = []

for ch in channels:
    cols = [c for c in feature_cols if c.startswith(ch + "_")]
    X_ch = df_dino[cols].to_numpy()

    norms = np.linalg.norm(X_ch, axis=1, keepdims=True)
    X_ch_l2 = X_ch / norms

    X_l2_parts.append(X_ch_l2)

X_l2 = np.concatenate(X_l2_parts, axis=1)

print(X_raw.shape)
print(X_l2.shape)
```

```python
# Check if normalized properly
start = 0

for ch in channels:
    X_ch_l2 = X_l2[:, start:start+768]
    norms = np.linalg.norm(X_ch_l2, axis=1)

    print(
        ch,
        np.mean(norms),
        np.min(norms),
        np.max(norms)
    )

    start += 768
```

```python
# PCA with 200 PCs
from sklearn.decomposition import PCA

pca_raw = PCA(n_components=200, random_state=42)
X_pca_raw = pca_raw.fit_transform(X_raw)

pca_l2 = PCA(n_components=200, random_state=42)
X_pca_l2 = pca_l2.fit_transform(X_l2)
```

```python
print(
    "Raw PCA variance explained by 200 PCs:",
    pca_raw.explained_variance_ratio_.sum()
)

print(
    "L2 PCA variance explained by 200 PCs:",
    pca_l2.explained_variance_ratio_.sum()
)
```

```python
for name, pca in [("raw", pca_raw), ("L2", pca_l2)]:
    cumulative = np.cumsum(pca.explained_variance_ratio_)

    print("\n", name)

    for n in [10, 20, 50, 100, 150, 200]:
        print(
            n,
            f"{cumulative[n-1]:.3f}"
        )
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(7, 5))

plt.plot(
    np.arange(1, 201),
    np.cumsum(pca_raw.explained_variance_ratio_),
    label="Raw"
)

plt.plot(
    np.arange(1, 201),
    np.cumsum(pca_l2.explained_variance_ratio_),
    label="Channel-wise L2"
)

plt.xlabel("Number of PCs")
plt.ylabel("Cumulative explained variance")
plt.legend()
plt.show()
```

```python
# no big difference between raw vs. L2 norm (variance explanation & cumulative explanation)
# -> 이건 현재 단계에서는 channel norm을 1로 맞추더라도 데이터의 global variance structure가 크게 바뀌지 않는다
# 90% variance 보존에 plate effect도 포함되어있을 수 있으므로 유의
```

# Quantitative QC
quantify replicate reproducibility: Do replicates show similar effects? Do we really need plate correction?

```python
# Select data with Rep = 2
condition_cols = [
    "Metadata_Compound",
    "Metadata_Concentration"
]

rep2_conditions = (
    df_dino.groupby(condition_cols)
    .size()
    .reset_index(name="n")
)

rep2_conditions = rep2_conditions[rep2_conditions["n"] == 2]
```

```python
df_dino = df_dino.reset_index(drop=True)
df_dino["row_id"] = np.arange(len(df_dino))
```

```python
rep2 = df_dino.merge(
    rep2_conditions[condition_cols],
    on=condition_cols,
    how="inner"
).copy()
```

```python
# Calculate cosine similarity between the two replicates from the same condition
from sklearn.metrics.pairwise import cosine_similarity

replicate_sims = []

for _, g in rep2.groupby(condition_cols):
    i, j = g["row_id"].to_numpy()

    sim = cosine_similarity(
        X_l2[[i]],
        X_l2[[j]]
    )[0, 0]

    replicate_sims.append(sim)

replicate_sims = np.array(replicate_sims)

print("n replicate pairs:", len(replicate_sims))
print("mean:", replicate_sims.mean())
print("median:", np.median(replicate_sims))
print("25/75%:", np.percentile(replicate_sims, [25, 75]))
```

```python
# Calculate cosine similarity from random pairs
rng = np.random.default_rng(42)

treatment_mask = df_dino["Metadata_Compound"] != "DMSO"
treatment_ids = df_dino.loc[treatment_mask, "row_id"].to_numpy()

n_pairs = len(replicate_sims)

random_sims = []

for _ in range(n_pairs):
    i, j = rng.choice(treatment_ids, size=2, replace=False)

    sim = cosine_similarity(
        X_l2[[i]],
        X_l2[[j]]
    )[0, 0]

    random_sims.append(sim)

random_sims = np.array(random_sims)

print("Random mean:", random_sims.mean())
print("Random median:", np.median(random_sims))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))
plt.hist(replicate_sims, bins=50, alpha=0.5, density=True, label="Replicates")
plt.hist(random_sims, bins=50, alpha=0.5, density=True, label="Random pairs")

plt.xlabel("Cosine similarity")
plt.ylabel("Density")
plt.legend()
plt.title("Replicate vs random similarity")
plt.show()
```

```python
# DINO embedding에는 강한 shared component가 있어서 모든 well 사이 cosine similarity가 매우 높고,
# uncentered cosine similarity만으로는 replicate reproducibility를 평가하기 어렵다.
# 그렇다고 representation이 나쁘다는 뜻은 아님 - PCA에서는 treatment-specific extreme phenotype이 명확하게 보였고,
# replicate distribution도 random보다 약간 오른쪽으로 이동함. 단지 측정 방법을 더 적절하게 바꿔야: source-pair matched random comparison
# 그다음 동일한 metric을 이용해서 uncorrected → plate-DMSO-centered가 얼마나 개선되는지 비교. 이후에 plate correction 여부 결정하기.
```

# source-pair matched random comparison
예. 실제 replicate가 prod_25 ↔ prod_27 조합이면 random pair도 똑같이 prod_25 ↔ prod_27에서 뽑는 방식

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

condition_cols = [
    "Metadata_Compound",
    "Metadata_Concentration"
]

# 실제 replicate pair 정보 만들기
rep_pairs = []

for _, g in rep2.groupby(condition_cols):
    if len(g) != 2:
        continue

    a = g.iloc[0]
    b = g.iloc[1]

    rep_pairs.append({
        "row_i": int(a["row_id"]),
        "row_j": int(b["row_id"]),
        "source_i": a["Metadata_source"],
        "source_j": b["Metadata_source"]
    })
```

```python
replicate_sims_matched = []

for p in rep_pairs:
    sim = cosine_similarity(
        X_l2[[p["row_i"]]],
        X_l2[[p["row_j"]]]
    )[0, 0]

    replicate_sims_matched.append(sim)

replicate_sims_matched = np.array(replicate_sims_matched)
```

```python
rng = np.random.default_rng(42)

# DMSO 제외
treatment_df = df_dino[
    df_dino["Metadata_Compound"] != "DMSO"
].copy()

source_to_ids = {
    source: g["row_id"].to_numpy()
    for source, g in treatment_df.groupby("Metadata_source")
}

random_sims_matched = []

for p in rep_pairs:
    s1 = p["source_i"]
    s2 = p["source_j"]

    # 같은 source일 수도 있으니 처리
    if s1 != s2:
        i = rng.choice(source_to_ids[s1])
        j = rng.choice(source_to_ids[s2])
    else:
        i, j = rng.choice(
            source_to_ids[s1],
            size=2,
            replace=False
        )

    sim = cosine_similarity(
        X_l2[[i]],
        X_l2[[j]]
    )[0, 0]

    random_sims_matched.append(sim)

random_sims_matched = np.array(random_sims_matched)
```

```python
print("Replicates")
print("mean:", replicate_sims_matched.mean())
print("median:", np.median(replicate_sims_matched))
print("IQR:", np.percentile(replicate_sims_matched, [25, 75]))

print("\nMatched random")
print("mean:", random_sims_matched.mean())
print("median:", np.median(random_sims_matched))
print("IQR:", np.percentile(random_sims_matched, [25, 75]))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))

plt.hist(
    replicate_sims_matched,
    bins=50,
    alpha=0.5,
    density=True,
    label="Replicates"
)

plt.hist(
    random_sims_matched,
    bins=50,
    alpha=0.5,
    density=True,
    label="Source-matched random"
)

plt.xlabel("Cosine similarity")
plt.ylabel("Density")
plt.title("Replicate vs source-matched random similarity")
plt.legend()
plt.show()
```

```python
diff = replicate_sims_matched - random_sims_matched

print("Mean similarity advantage:", diff.mean())
print(
    "Fraction replicate > matched random:",
    np.mean(diff > 0)
)
```

```python
# 여전히 random pair와 replicate간의 유사성이 크게 다르지 않음. baseline effect가 크다고 판단됨.
```

# Plate-specific DMSO centering

```python
# Calculate plate-specific DMSO centriod
import numpy as np

df_dino = df_dino.reset_index(drop=True)

# row alignment 확인용
df_dino["row_id"] = np.arange(len(df_dino))

# plate 목록
plates = df_dino["Metadata_Plate"].unique()

# corrected embedding 저장
X_l2_dmso_centered = np.zeros_like(X_l2)

for plate in plates:
    plate_mask = df_dino["Metadata_Plate"].eq(plate)
    dmso_mask = (
        plate_mask &
        df_dino["Metadata_Compound"].eq("DMSO")
    )

    dmso_ids = df_dino.loc[dmso_mask, "row_id"].to_numpy()
    plate_ids = df_dino.loc[plate_mask, "row_id"].to_numpy()

    if len(dmso_ids) == 0:
        print("No DMSO on plate:", plate)
        X_l2_dmso_centered[plate_ids] = X_l2[plate_ids]
        continue

    dmso_centroid = X_l2[dmso_ids].mean(axis=0)

    X_l2_dmso_centered[plate_ids] = (
        X_l2[plate_ids] - dmso_centroid
    )
```

```python
# Check if all plates have DMSO
dmso_per_plate = (
    df_dino[df_dino["Metadata_Compound"] == "DMSO"]
    .groupby("Metadata_Plate")
    .size()
)

print(dmso_per_plate.describe())

missing_dmso_plates = set(plates) - set(dmso_per_plate.index)

print("Plates without DMSO:", missing_dmso_plates)
```

```python
# Sanity check - check if correction is done properly
check = []

for plate in plates:
    dmso_mask = (
        df_dino["Metadata_Plate"].eq(plate) &
        df_dino["Metadata_Compound"].eq("DMSO")
    )

    ids = df_dino.loc[dmso_mask, "row_id"].to_numpy()

    if len(ids) == 0:
        continue

    centered_mean = X_l2_dmso_centered[ids].mean(axis=0)

    check.append(
        np.linalg.norm(centered_mean)
    )

print("Mean norm of plate DMSO centroids after correction:",
      np.mean(check))

print("Max:",
      np.max(check))
```

```python
# Calculate replicate similarity
from sklearn.metrics.pairwise import cosine_similarity

replicate_sims_centered = []

for p in rep_pairs:
    sim = cosine_similarity(
        X_l2_dmso_centered[[p["row_i"]]],
        X_l2_dmso_centered[[p["row_j"]]]
    )[0, 0]

    replicate_sims_centered.append(sim)

replicate_sims_centered = np.array(replicate_sims_centered)
```

```python
# Calculate source-matched random
rng = np.random.default_rng(42)

random_sims_centered = []

for p in rep_pairs:
    s1 = p["source_i"]
    s2 = p["source_j"]

    if s1 != s2:
        i = rng.choice(source_to_ids[s1])
        j = rng.choice(source_to_ids[s2])
    else:
        i, j = rng.choice(
            source_to_ids[s1],
            size=2,
            replace=False
        )

    sim = cosine_similarity(
        X_l2_dmso_centered[[i]],
        X_l2_dmso_centered[[j]]
    )[0, 0]

    random_sims_centered.append(sim)

random_sims_centered = np.array(random_sims_centered)
```

```python
import numpy as np
from sklearn.decomposition import PCA

def fit_sphering(X_control, n_components=None, eps=1e-6):
    """
    Fit PCA/ZCA-like whitening using negative controls.
    Returns control mean, PCA object, and scale factors.
    """
    mu = X_control.mean(axis=0)
    Xc = X_control - mu

    if n_components is None:
        n_components = min(Xc.shape[0] - 1, Xc.shape[1])

    pca = PCA(
        n_components=n_components,
        svd_solver="randomized",
        random_state=42
    )
    pca.fit(Xc)

    scales = 1.0 / np.sqrt(pca.explained_variance_ + eps)

    return mu, pca, scales
```

```python
def apply_sphering(X, mu, pca, scales):
    Xc = X - mu

    X_pc = pca.transform(Xc)
    X_white = X_pc * scales

    return X_white
```

```python
# 1) Raw L2 에 global sphering
mu_global, pca_global, scales_global = fit_sphering(
    X_dmso,
    n_components=500,
    eps=1e-6
)

X_l2_sphered = apply_sphering(
    X_l2,
    mu_global,
    pca_global,
    scales_global
)

print(X_l2_sphered.shape)
```

```python
# 2) plate-centered L2에 global sphering
mu_pc, pca_pc, scales_pc = fit_sphering(
    X_dmso_centered,
    n_components=500,
    eps=1e-6
)

X_l2_centered_sphered = apply_sphering(
    X_l2_dmso_centered,
    mu_pc,
    pca_pc,
    scales_pc
)

print(X_l2_centered_sphered.shape)
```

```python
def evaluate_replicates(
    X,
    rep_pairs,
    source_to_ids,
    n_random=100,
    seed=42
):
    rng = np.random.default_rng(seed)

    percentiles = []

    for p in rep_pairs:
        i_true = p["row_i"]
        j_true = p["row_j"]

        s1 = p["source_i"]
        s2 = p["source_j"]

        # true replicate similarity
        x1 = X[i_true]
        x2 = X[j_true]

        true_sim = np.dot(x1, x2) / (
            np.linalg.norm(x1) *
            np.linalg.norm(x2)
        )

        random_sims = []

        for _ in range(n_random):

            if s1 != s2:
                i = rng.choice(source_to_ids[s1])
                j = rng.choice(source_to_ids[s2])

            else:
                i, j = rng.choice(
                    source_to_ids[s1],
                    size=2,
                    replace=False
                )

            a = X[i]
            b = X[j]

            sim = np.dot(a, b) / (
                np.linalg.norm(a) *
                np.linalg.norm(b)
            )

            random_sims.append(sim)

        random_sims = np.array(random_sims)

        percentile = (
            np.mean(true_sim > random_sims) * 100
        )

        percentiles.append(percentile)

    percentiles = np.array(percentiles)

    return {
        "median": np.median(percentiles),
        "mean": np.mean(percentiles),
        "pct_ge_90": np.mean(percentiles >= 90),
        "pct_ge_95": np.mean(percentiles >= 95),
        "percentiles": percentiles
    }
```

```python
results = {}

results["L2"] = evaluate_replicates(
    X_l2,
    rep_pairs,
    source_to_ids
)

results["Plate centered"] = evaluate_replicates(
    X_l2_dmso_centered,
    rep_pairs,
    source_to_ids
)

results["Global sphered"] = evaluate_replicates(
    X_l2_sphered,
    rep_pairs,
    source_to_ids
)

results["Plate centered + sphered"] = evaluate_replicates(
    X_l2_centered_sphered,
    rep_pairs,
    source_to_ids
)
```

```python
# percentile evaluation by concentration
def replicate_percentiles_by_concentration(
    X,
    rep_pairs,
    source_to_ids,
    n_random=100,
    seed=42
):
    rng = np.random.default_rng(seed)
    rows = []

    for p in rep_pairs:
        i_true = p["row_i"]
        j_true = p["row_j"]
        s1 = p["source_i"]
        s2 = p["source_j"]

        x1 = X[i_true]
        x2 = X[j_true]

        true_sim = np.dot(x1, x2) / (
            np.linalg.norm(x1) * np.linalg.norm(x2)
        )

        random_sims = []

        for _ in range(n_random):
            if s1 != s2:
                i = rng.choice(source_to_ids[s1])
                j = rng.choice(source_to_ids[s2])
            else:
                i, j = rng.choice(
                    source_to_ids[s1],
                    size=2,
                    replace=False
                )

            a = X[i]
            b = X[j]

            sim = np.dot(a, b) / (
                np.linalg.norm(a) * np.linalg.norm(b)
            )

            random_sims.append(sim)

        random_sims = np.array(random_sims)

        percentile = np.mean(true_sim > random_sims) * 100

        rows.append({
            "concentration": p["concentration"],
            "percentile": percentile,
            "true_similarity": true_sim
        })

    return pd.DataFrame(rows)
```

```python
# 단순 count만 보면 원래 데이터에 많이 존재하는 pathway가 당연히 많이 나오니까, enrichment ratio를 확인하기.
def category_enrichment(subset, background, col, min_background=10):
    bg = background[col].value_counts(dropna=False)
    sub = subset[col].value_counts(dropna=False)

    out = pd.DataFrame({
        "subset_n": sub,
        "background_n": bg
    }).fillna(0)

    out = out[out["background_n"] >= min_background]

    out["subset_frac"] = out["subset_n"] / len(subset)
    out["background_frac"] = out["background_n"] / len(background)

    out["fold_enrichment"] = (
        out["subset_frac"] / out["background_frac"]
    )

    return out.sort_values(
        ["fold_enrichment", "subset_n"],
        ascending=False
    )
```

# Exploratory UMAP

```python
# 네 상황처럼 DINO embedding + L2 normalization + plate별 DMSO centering이라면 저는 umap metric에서 cosine과 euclidean을 둘 다 확인하는 걸 추천해요.

# metric="euclidean": embedding에서 변화의 방향 + 크기가 모두 반영됩니다. DMSO에서 많이 벗어난 strong phenotype이 거리에도 크게 반영돼요.
# metric="cosine": 주로 변화 방향을 봅니다. DMSO 대비 phenotype 변화가 같은 방향이면 effect magnitude가 달라도 가까워질 수 있어요.
# metric="correlation": classical Cell Painting feature/profile에서는 꽤 자연스러운 선택입니다. feature pattern의 상대적 모양을 비교하기 때문이에요. 다만 DINO embedding에서는 보통 cosine을 먼저 고려할 것 같아요.

# “어떤 방향의 morphological phenotype인가?” → cosine
# “phenotype 방향뿐 아니라 effect strength까지 비슷한가?” → euclidean

# 그리고 중요한 게 하나 있어요. PCA를 먼저 했다면 보통은 PCA → UMAP(metric="euclidean") 조합이 매우 흔합니다. PCA 공간 자체가 Euclidean geometry를 기반으로 만들어진 공간이라 해석도 자연스럽고요.
# 반면 네가 PCA 없이 DINO embedding 자체를 바로 넣는다면, coosine 을 확인하는걸 추천.

# “어느 UMAP 그림이 더 예쁜가”보다 같은 compound의 biological replicate가 잘 모이는지, known MoA가 가까워지는지, plate별 cluster가 생기지 않는지를 기준으로 고르는 게 더 중요합니다.
# 특히 지금처럼 DMSO centering을 했다면 개인적으로는 centered DINO → cosine UMAP과 PCA → euclidean UMAP을 먼저 비교해볼 것 같아요. 이 둘의 결과 차이를 보면 네 embedding에서 direction이 중요한지, magnitude까지 중요한지도 꽤 잘 드러납니다.
```

# plate별 DMSO를 기준으로 cytotoxicity flag

```python
# 각 plate의 DMSO 분포에서 median과 MAD를 기준으로 z-score
import numpy as np
import pandas as pd

tox_cols = [
    "Metadata_Count_Cells",
    "Metadata_mtt_normalized",
    "Metadata_ldh_normalized"
]

def mad(x):
    med = np.nanmedian(x)
    return np.nanmedian(np.abs(x - med))

plate_stats = {}

for plate, g in df_dino[
    df_dino["Metadata_Compound"].eq("DMSO")
].groupby("Metadata_Plate"):

    plate_stats[plate] = {}

    for col in tox_cols:
        vals = g[col].to_numpy()

        plate_stats[plate][col] = {
            "median": np.nanmedian(vals),
            "mad": mad(vals)
        }
```

```python
# 각 well이 자기 plate의 DMSO에서 얼마나 벗어났는지 계산
for col in tox_cols:
    df_dino[col + "_dmso_z"] = np.nan

for idx, row in df_dino.iterrows():
    plate = row["Metadata_Plate"]

    if plate not in plate_stats:
        continue

    for col in tox_cols:
        med = plate_stats[plate][col]["median"]
        m = plate_stats[plate][col]["mad"]

        if np.isnan(m) or m == 0:
            continue

        z = (row[col] - med) / (1.4826 * m)

        df_dino.loc[idx, col + "_dmso_z"] = z
```

```python
# baseline: 3 MAD-equivalent SD
df_dino["cytotoxic_flag"] = (
    (df_dino["Metadata_Count_Cells_dmso_z"] < -3)
    |
    (df_dino["Metadata_mtt_normalized_dmso_z"] < -3)
    |
    (df_dino["Metadata_ldh_normalized_dmso_z"] > 3)
)
```

```python
from sklearn.decomposition import PCA

pca_nt = PCA(
    n_components=50,
    random_state=42
)

X_pca50_nt = pca_nt.fit_transform(
    X_l2_sphered[non_toxic_idx]
)

print(X_pca50_nt.shape)
print(
    "Variance explained:",
    pca_nt.explained_variance_ratio_.sum()
)
```

```python
# UMAP calculation
import umap.umap_ as umap

umap_nt = umap.UMAP(
    n_neighbors=30,
    min_dist=0.1,
    metric="euclidean",
    random_state=42
)

X_umap_nt = umap_nt.fit_transform(
    X_pca50_nt
)

print(X_umap_nt.shape)
```

# reproducible replicate-condition level

```python
condition_cols = [
    "Metadata_Compound",
    "Metadata_Concentration"
]

# strict non-toxic wells
strict_mask = df_dino["n_tox_flags"].eq(0)

# condition별 전체 replicate 수 / strict non-toxic replicate 수
cond_qc = (
    df_dino
    .groupby(condition_cols)
    .agg(
        n_total=("row_id", "size"),
        n_strict=("n_tox_flags", lambda x: (x == 0).sum())
    )
    .reset_index()
)

# 원래 replicate가 정확히 2개이고,
# 두 well 모두 strict non-toxic인 condition
good_conditions = cond_qc[
    (cond_qc["n_total"] == 2) &
    (cond_qc["n_strict"] == 2)
][condition_cols]

print("Usable replicate conditions:", len(good_conditions))
```

```python
good_rows = (
    df_dino
    .reset_index()
    .merge(
        good_conditions,
        on=condition_cols,
        how="inner"
    )
)

# 원래 df_dino의 위치
row_pos = good_rows["index"].to_numpy()

# 각 condition을 integer group으로
cond_code, cond_unique = pd.factorize(
    list(zip(
        good_rows["Metadata_Compound"],
        good_rows["Metadata_Concentration"]
    ))
)

n_cond = len(cond_unique)

# PCA 전에 replicate 평균
X_cond = np.zeros(
    (n_cond, X_l2_sphered.shape[1]),
    dtype=np.float32
)

for i in range(n_cond):
    idx = row_pos[cond_code == i]
    X_cond[i] = X_l2_sphered[idx].mean(axis=0)
```

```python
pca_cond = PCA(
    n_components=50,
    random_state=42
)

X_cond_pca50 = pca_cond.fit_transform(X_cond)

umap_cond = umap.UMAP(
    n_neighbors=30,
    min_dist=0.1,
    metric="euclidean",
    random_state=42
)

X_cond_umap = umap_cond.fit_transform(X_cond_pca50)
```

```python
print(X_cond_umap.shape)

plt.figure(figsize=(8, 7))
plt.scatter(
    X_cond_umap[:, 0],
    X_cond_umap[:, 1],
    s=10,
    alpha=0.5
)

plt.xlabel("UMAP1")
plt.ylabel("UMAP2")
plt.title("Condition-level strict non-toxic DINO UMAP")
plt.show()
```
